# Spintegrate — GUI
Этот ноутбук запускает обновлённое приложение. Держите его рядом с папкой `spintegrate`.
Исходный алгоритм: локальная коррекция базовой линии → peak-picking → подгонка Лоренцев с фиксированными центрами → интеграл `Σ A × HWHM`.
Актуальная численная реализация находится в `spintegrate/core.py`. Исходный исследовательский ноутбук с сохранёнными результатами остаётся в локальной копии проекта.

```python
# Запустить отдельное GUI-окно в Python текущего ядра.
# Ячейка не блокируется; окно можно закрыть и запустить снова.
import subprocess
import sys
from pathlib import Path
app_path = Path.cwd() / 'spintegrate_app.py'
if not app_path.exists():
    raise FileNotFoundError('Откройте ноутбук из папки проекта Spintegrate.')
gui_process = subprocess.Popen([sys.executable, str(app_path)], cwd=str(app_path.parent))
print('Окно Spintegrate запущено. PID:', gui_process.pid)
```
